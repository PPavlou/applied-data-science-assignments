```python
from datasets import load_dataset
from sentence_transformers import SentenceTransformer
from sklearn.cluster import KMeans
import numpy as np
import random
import re
import pandas as pd
import difflib
```

```python
ds = load_dataset("DominusTea/GreekLegalSum", split="train")
texts, summaries = ds["text"], ds["summary"]
```

```python
input_texts = [
    summ.strip() if summ and summ.strip() else txt[:2000]
    for txt, summ in zip(texts, summaries)
]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

tfidf = TfidfVectorizer(max_features=5000, ngram_range=(1,2), min_df=5)
X = tfidf.fit_transform(input_texts)

K = 10
kmeans = KMeans(n_clusters=K, random_state=42).fit(X)
labels, centroids = kmeans.labels_, kmeans.cluster_centers_

cluster_indices = {i: np.where(labels==i)[0].tolist() for i in range(K)}
```

```python
random_examples, centroid_examples = {}, {}
for clus, idxs in cluster_indices.items():
    random_examples[clus] = random.sample(idxs, min(3, len(idxs)))
    chunk = X[idxs].toarray()
    cent = centroids[clus]
    dists = np.linalg.norm(chunk - cent, axis=1)
    top3 = np.argsort(dists)[:3]
    centroid_examples[clus] = [idxs[i] for i in top3]
```

```python
def build_prompt(example_ids, target_id):
    shots = ""
    for eid in example_ids:
        shots += f"Παράδειγμα:\n{input_texts[eid]}\n\n"
    return (
        "Σου δίνεται ένα κείμενο νομικής απόφασης. Ποιο είναι το κεντρικό θέμα της απόφασης;\n\n"
         + shots
         + "Κείμενο προς ανάλυση:\n"
         + input_texts[target_id]
         + "\n\n"
         +"Απάντησε ακριβώς με:\nΘέμα: <τίτλος σε έως 10 λέξεις>\n"
    )
```

```python
import re

def extract_title(resp: str) -> str:

    matches = re.findall(r"Θέμα:\s*(.+?)(?:\r?\n|$)", resp)

    real = [m.strip() for m in matches if not m.strip().startswith("<τίτλος")]
    if not real:
        return ""

    title = real[-1].rstrip(".").strip()
    return title


records = []
for method, examples in [("random", random_examples), ("centroid", centroid_examples)]:
    for clus, ex_ids in examples.items():
        target = cluster_indices[clus][0]
        prompt = build_prompt(ex_ids, target)


        print(f"\n\n=== DEBUG CLUSTER {clus} ({method}) PROMPT ===")
        print(prompt)


        msgs = [
            {"role": "system", "content": "Είσαι ένα σύστημα ΤΝ για ανάλυση νομικών αποφάσεων."},
            {"role": "user",   "content": prompt}
        ]
        inputs = tokenizer.apply_chat_template(
            msgs, tokenize=True, add_generation_prompt=True, return_tensors="pt"
        ).to(device)
        attention_mask = torch.ones_like(inputs, device=device)


        out_ids = model.generate(
            input_ids=inputs,
            attention_mask=attention_mask,
            max_new_tokens=64,
            use_cache=True
        )
        resp = tokenizer.decode(out_ids[0], skip_special_tokens=True)


        print(f"\n--- RAW MODEL RESPONSE ({clus}, {method}) ---")
        print(resp)


        lines = resp.splitlines()
        for i, line in enumerate(lines):
            if "Θέμα:" in line:
                print(f"\n(line {i}) {line}")


        matches = re.findall(r"Θέμα:\s*(.+?)(?:\r?\n|$)", resp)
        print("\nREGEX FINDALL MATCHES:", matches)


        title = extract_title(resp)
        print("\nEXTRACTED TITLE:", repr(title))
        print(f"[{clus}/{method}] all matches:", re.findall(r"Θέμα:\s*(.+?)(?:\r?\n|$)", resp))
        print(f"[{clus}/{method}] extracted title:", title)

        human = summaries[target].strip()
        copied_exactly = (title == human)
        sim_ratio = difflib.SequenceMatcher(None, title, human).ratio()

        records.append({
            "cluster": clus,
            "method": method,
            "generated_title": title,
            "copied_exactly": copied_exactly,
            "similarity_to_summary": round(sim_ratio, 2)
        })
```

```python
df = pd.DataFrame(records)
print("\nExact copies per method (should be 0):")
print(df.groupby("method")["copied_exactly"].sum())
print("\nAverage similarity to summary (lower is better):")
print(df.groupby("method")["similarity_to_summary"].mean())
display(df.head(20))
```
